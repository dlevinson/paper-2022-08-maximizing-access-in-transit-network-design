### Stops

```python
# Join

print(len(stops_current))
print(len(stops_rail))
print(len(stops_road))
print(len(stops_mg))
stops_combined = stops_current.append(stops_rail, ignore_index=True).append(stops_road, ignore_index=True).append(stops_mg, ignore_index=True)
print(len(stops_combined))
```

```python
# Drop duplicates (parent stations would be repeating)

print(len(stops_combined))
stops_combined['stop_id'] = stops_combined['stop_id'].astype(str)
stops_combined['stop_name'] = stops_combined['stop_name'].astype(str)
stops_combined = stops_combined.drop_duplicates(subset=['stop_id','stop_name'], keep='first')
print(len(stops_combined))
```

```python
# Edit data types (to avoid errors during validation)

stops_combined['location_type'] = stops_combined['location_type'].apply(lambda x: str(int(x)) if x>=0 else '')
stops_combined['location_type'].unique()
```
