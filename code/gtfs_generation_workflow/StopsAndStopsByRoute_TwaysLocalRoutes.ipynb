### Combine to and from route distances for local routes

First remove any stops that need to be removed (especially in Grid and 1TOD scenarios)

```python
stopsseparate = pd.read_csv('./Stops/stopsByRoute_TwaysLocalRoutes_withDistAlongR_TOD14.csv')
stopsseparate.head()
```

```python
len(stopsseparate)
```

```python
shapesforstops = pd.read_csv('./Shapes/twayLocalRouteShapes_forStops_TOD14.csv')
shapesforstops.head()
```

#### Get parent routes with to_route lengths

```python
localroutes = shapesforstops.loc[(shapesforstops['shape_id'].str.contains('Austral')) | (shapesforstops['shape_id'].str.contains('Rossmore')) | (shapesforstops['shape_id'].str.contains('KempsCreek'))]
localroutes = localroutes[['shape_id','shape_dist_traveled']].groupby('shape_id', as_index=False).max()
len(localroutes)
```

```python
localroutes['parent_route'] = localroutes['shape_id'].str.rstrip('-from').str.rstrip('-to')
localroutes.head()
```

```python
parentroutes = localroutes.loc[localroutes['shape_id'].str.contains('-to')].drop('shape_id',axis=1)
parentroutes.columns = ['toroute_length','parent_route']
print(len(parentroutes))
parentroutes.head()
```

#### Separate tway and local route stops

```python
stopsseparate['parent_route'] = stopsseparate['shape_id'].str.rstrip('-from').str.rstrip('-to')
stopsseparate.head()
```

```python
stops_tways = stopsseparate.loc[(stopsseparate['shape_id'].str.contains('Austral')==False) & (stopsseparate['shape_id'].str.contains('Rossmore')==False) & (stopsseparate['shape_id'].str.contains('KempsCreek')==False)]
stops_fromroutes = stopsseparate.loc[stopsseparate['shape_id'].str.contains('-from')]
stops_toroutes = stopsseparate.loc[stopsseparate['shape_id'].str.contains('-to')]
print(len(stops_tways))
print(len(stops_fromroutes))
print(len(stops_toroutes))
```

```python
stops_tways['shape_id'].unique()
```

#### Add to_route length to from_route stop distances to combine to and from routes into one route

```python
print(len(stops_fromroutes))
stops_fromroutes = stops_fromroutes.merge(parentroutes, how='left', on='parent_route')
print(len(stops_fromroutes))
stops_fromroutes.head()
```

```python
stops_fromroutes.isnull().sum()
```

```python
stops_fromroutes['distAlongR'].sum() + stops_fromroutes['toroute_length'].sum()
```

```python
stops_fromroutes['distAlongR'] = stops_fromroutes['distAlongR']+stops_fromroutes['toroute_length']
stops_fromroutes['distAlongR'].sum()
```

```python
stops_fromroutes = stops_fromroutes.drop('toroute_length', axis=1)
stops_fromroutes.head()
```

### Merge tway stops, from route and to route stops, and generate stops.txt

```python
(stops_tways['shape_id']!=stops_tways['parent_route']).sum()
```

```python
# StopsByRoute for stoptimes
stops = stops_tways.append(stops_toroutes).append(stops_fromroutes).sort_values(['parent_route','distAlongR'])
stops.to_csv('StopsByRoute_TwaysAndLocalRoutes_TOD14.csv', index=False)
```

```python
# stops.txt
stops['stop_id'] = stops['stop_id'].astype(str)
stops['stop_code'] = stops['stop_id'].astype(float)
stops['wheelchair_boarding'] = "0"
stops['location_type'] = ""
stops['parent_station'] = ""
stops['platform_code'] = ""
stops = stops[['stop_id','stop_code','stop_name','stop_lat','stop_lon','location_type','parent_station','wheelchair_boarding','platform_code']]
stops.head()
```
