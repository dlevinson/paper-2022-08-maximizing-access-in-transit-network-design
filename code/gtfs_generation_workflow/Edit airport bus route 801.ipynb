### Remove stops on Badgerys Creek Road

```python
# Remove stops from stops

print(len(stops))
stops = stops.loc[stops['stop_id'].isin(stopidstoremove)==False]
print(len(stops))
```

```python
# Remove stops from stop_times

print(len(stoptimes))
stoptimes = stoptimes.loc[stoptimes['stop_id'].isin(stopidstoremove)==False]
print(len(stoptimes))
```

```python
# Adjust stop sequence of corresponding trips

stoptimes_trips801 = stoptimes.loc[stoptimes['trip_id'].isin(tripids801)]
len(stoptimes_trips801)
```

```python
stoptimes_trips801['stop_sequence'] = stoptimes_trips801.groupby('trip_id')['shape_dist_traveled'].rank(ascending=True).astype(int)
stoptimes_trips801
```

```python
print(len(stoptimes))
stoptimes = stoptimes.loc[stoptimes['trip_id'].isin(tripids801)==False]
print(len(stoptimes))
```

```python
stoptimes = stoptimes.append(stoptimes_trips801)
```
