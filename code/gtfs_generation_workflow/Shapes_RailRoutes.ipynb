```python
# Train route T2 extension (City - Leppington - WS Aerotropolis)

routeT2h = routeShapes.loc[routeShapes['shape_id']=='2-T2-sj2-22.76.H'].sort_values('vertex_ind') # City circle to Leppington 
routeT2h = routeT2h.append(routeShapes.loc[routeShapes['shape_id']=='LP-C-B2'].sort_values('vertex_ind')) # Leppington to Aerotropolis connection
routeT2h = routeT2h.append(routeShapes.loc[routeShapes['shape_id']=='M-SW-RL-2'].sort_values('vertex_ind')) # Leppington to Aerotropolis
routeT2h = routeT2h.reset_index(drop=True)
routeT2h['shape_pt_sequence'] = routeT2h.index+1
routeT2h['shape_id'] = 'T2-CTYC-WSAR'

routeT2r = routeShapes.loc[routeShapes['shape_id']=='M-SW-RL-1'].sort_values('vertex_ind') # Aerotropolis to Leppington
routeT2r = routeT2r.append(routeShapes.loc[routeShapes['shape_id']=='LP-C-B1'].sort_values('vertex_ind')) # Aerotropolis to Leppington connection
routeT2r = routeT2r.append(routeShapes.loc[routeShapes['shape_id']=='2-T2-sj2-22.9.R'].sort_values('vertex_ind')) # Leppington to City circle
routeT2r = routeT2r.reset_index(drop=True)
routeT2r['shape_pt_sequence'] = routeT2r.index+1
routeT2r['shape_id'] = 'T2-WSAR-CTYC'

routeT2 = routeT2h.append(routeT2r)
routeT2 = routeT2[['shape_id','shape_lat','shape_lon','shape_pt_sequence','distance']]
routeT2
```

```python
futureRailRouteShapes = route1.append(route2)
futureRailRouteShapes = futureRailRouteShapes.append(route3a)
futureRailRouteShapes = futureRailRouteShapes.append(route3b)
futureRailRouteShapes = futureRailRouteShapes.append(route4a)
futureRailRouteShapes = futureRailRouteShapes.append(route4b)
futureRailRouteShapes = futureRailRouteShapes.append(route5)
futureRailRouteShapes = futureRailRouteShapes.append(route6)
futureRailRouteShapes = futureRailRouteShapes.append(routePLR)
futureRailRouteShapes = futureRailRouteShapes.append(routeT2)
futureRailRouteShapes.reset_index(drop=True, inplace=True)
futureRailRouteShapes['shape_id'].unique()
```

```python
futureRailRouteShapes = futureRailRouteShapes.sort_values(['shape_id','shape_pt_sequence'], ascending=[True,True])
futureRailRouteShapes['geopy_dist'] = np.nan
for i in range(len(futureRailRouteShapes)):
    if futureRailRouteShapes['shape_pt_sequence'][i] == 1:
        futureRailRouteShapes['geopy_dist'][i] = 0
    else:
        p1 = (futureRailRouteShapes['shape_lat'][i-1],futureRailRouteShapes['shape_lon'][i-1]) 
        p2 = (futureRailRouteShapes['shape_lat'][i],futureRailRouteShapes['shape_lon'][i])
        futureRailRouteShapes['geopy_dist'][i] = geo.sphere.haversine_distance(p1,p2)
futureRailRouteShapes['cum_dist'] = futureRailRouteShapes.groupby(['shape_id'])['geopy_dist'].cumsum()
futureRailRouteShapes.head()
```

```python
futureRailRouteShapes.drop(['distance','geopy_dist'], axis=1, inplace=True)
futureRailRouteShapes.columns = ['shape_id','shape_pt_lat','shape_pt_lon','shape_pt_sequence','shape_dist_traveled']
futureRailRouteShapes.head()
```

```python
futureRailRouteShapes.to_csv('./output/future/shapes.txt', index=False)
```

```python
# Current GTFS

shapes = pd.read_csv('./input/gtfs_static_8jul2020/shapes.txt')
routes = pd.read_csv('./input/gtfs_static_8jul2020/routes.txt')
trips = pd.read_csv('./input/gtfs_static_8jul2020/trips.txt')
stoptimes = pd.read_csv('./input/gtfs_static_8jul2020/stop_times.txt')
calendar = pd.read_csv('./input/gtfs_static_8jul2020/calendar.txt')
calendardates = pd.read_csv('./input/gtfs_static_8jul2020/calendar_dates.txt')
stops = pd.read_csv('./input/gtfs_static_8jul2020/stops.txt')
```

```python
# Remove existing metro line (but keep stops)

print(len(routes['route_id']))
routes = routes.loc[routes['route_desc']!='Sydney Metro Network']
print(len(routes['route_id']))

print(len(trips))
trips['route_id'] = trips['route_id'].astype(str)
routes['route_id'] = routes['route_id'].astype(str)
trips = trips.loc[trips['route_id'].isin(routes['route_id'])]
print(len(trips))

print(len(stoptimes))
stoptimes['trip_id'] = stoptimes['trip_id'].astype(str)
trips['trip_id'] = trips['trip_id'].astype(str)
stoptimes = stoptimes.loc[stoptimes['trip_id'].isin(trips['trip_id'])]
print(len(stoptimes))

print(len(shapes))
shapes['shape_id'] = shapes['shape_id'].astype(str)
trips['shape_id'] = trips['shape_id'].astype(str)
shapes = shapes.loc[shapes['shape_id'].isin(trips['shape_id'])]
print(len(shapes))

print(len(calendar))
calendar['service_id'] = calendar['service_id'].astype(str)
trips['service_id'] = trips['service_id'].astype(str)
calendar = calendar.loc[calendar['service_id'].isin(trips['service_id'])]
print(len(calendar))

print(len(calendardates))
calendardates['service_id'] = calendardates['service_id'].astype(str)
trips['service_id'] = trips['service_id'].astype(str)
calendardates = calendardates.loc[calendardates['service_id'].isin(trips['service_id'])]
print(len(calendardates))
```
